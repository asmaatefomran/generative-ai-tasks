# **Task 7: Build a Multi-Agent AI System for Report Generation**

## Objective

Build a simple multi-agent AI system where Planner, Research, Writer,
and Summary agents collaborate to generate a structured report from
a user question.


User Question

 ↓

Planner Agent

 ↓

Research Agent

  ↓

Writer Agent

↓

Summary Agent

↓

Final Report

Setup Groq

```python
import os
from google.colab import userdata
from groq import Groq

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

Create a simple LLM function

```python
def ask_llm(prompt):

    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0
    )

    return response.choices[0].message.content
```

# Planner Agent

The Planner decides what the report should cover

```python
def planner_agent(question):

    prompt = f"""
You are the Planner Agent.

Create a simple plan for answering this question.

Question:
{question}

Give 3 to 5 important points that should be covered in the report.
"""

    return ask_llm(prompt)
```

# Research Agent

The Research Agent provides information about the planned topics

```python
def research_agent(question, plan):

    prompt = f"""
You are the Research Agent.

Research the following question based on your general knowledge.

Question:
{question}

Plan:
{plan}

Provide useful factual information for each point in the plan.
"""

    return ask_llm(prompt)
```

# Writer Agent

The Writer takes the research and creates the report

```python
def writer_agent(question, research):

    prompt = f"""
You are the Writer Agent.

Write a structured report answering the question.

Question:
{question}

Research:
{research}

Use:
- A clear title
- Introduction
- Main sections
- Conclusion

Do not invent facts that are not supported by the research.
"""

    return ask_llm(prompt)
```

# Summary Agent

The Summary Agent creates the final summary

```python
def summary_agent(report):

    prompt = f"""
You are the Summary Agent.

Read the report below and create a short summary.

Report:
{report}

Provide:
- 3 key points
- A short conclusion
"""

    return ask_llm(prompt)
```

Run the Multi-Agent System

```python
question = input("Enter your question: ")

print("\n--- Planner Agent ---")
plan = planner_agent(question)
print(plan)

print("\n--- Research Agent ---")
research = research_agent(question, plan)
print(research)

print("\n--- Writer Agent ---")
report = writer_agent(question, research)
print(report)

print("\n--- Summary Agent ---")
summary = summary_agent(report)
print(summary)
```

```python
print("\n" + "=" * 60)
print("FINAL REPORT")
print("=" * 60)

print(report)

print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)

print(summary)
```
