# **Task 6: Build a Context-Aware Chat Assistant with Memory**

## Objective

Build a simple AI chat assistant that maintains conversation history
and uses previous messages as context when generating new responses.

Imports and API key

```python
import os
from google.colab import userdata
from groq import Groq

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

Create conversation memory

```python
conversation_history = []
```

Create the chat function

```python
def chat(user_message):

    conversation_history.append(
        {
            "role": "user",
            "content": user_message
        }
    )

    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=conversation_history,
        temperature=0
    )

    assistant_message = response.choices[0].message.content

    conversation_history.append(
        {
            "role": "assistant",
            "content": assistant_message
        }
    )

    return assistant_message
```

Continuous chat loop

```python
while True:

    user_message = input("\nYou: ")

    if user_message.lower() == "exit":
        print("Chat ended.")
        break

    response = chat(user_message)

    print("Assistant:", response)
```
