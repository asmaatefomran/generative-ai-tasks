# **Task 8: Automated Generative AI Email Assistant with API Functions**
## Objective

Build a simple AI email assistant that uses an LLM to generate
professional customer replies and automatically processes them
through API-like functions for saving, sending, and logging.

Setup

```python
import os
from google.colab import userdata
from groq import Groq

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

# API-like functions
These simulate backend operations

# API Function 1 - Generate Email Reply
This represents an LLM API call

```python
def generate_reply(customer_message):

    if not customer_message.strip():
        return {
            "success": False,
            "error": "Customer message is empty."
        }

    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-120b",
            messages=[
                {
                    "role": "system",
                    "content": """
You are a professional customer support email assistant.
Generate a polite and helpful email reply.
Return only the email.
"""
                },
                {
                    "role": "user",
                    "content": customer_message
                }
            ],
            temperature=0
        )

        return {
            "success": True,
            "email": response.choices[0].message.content
        }

    except Exception as e:
        return {
            "success": False,
            "error": str(e)
        }
```

# API Function 2 - Save Email
This simulates a backend storage API

```python
def save_email(email):

    try:
        with open("emails.txt", "a") as file:
            file.write(email)
            file.write("\n" + "=" * 50 + "\n")

        return {
            "success": True,
            "message": "Email saved successfully."
        }

    except Exception as e:
        return {
            "success": False,
            "error": str(e)
        }
```

#API Function 3 - Send Email
This simulates an email-sending API


```python
def send_email(email, recipient="customer@example.com"):

    try:
        print("\n[EMAIL API]")
        print(f"Sending email to: {recipient}")
        print("Email sent successfully.")

        return {
            "success": True,
            "recipient": recipient,
            "message": "Email sent successfully."
        }

    except Exception as e:
        return {
            "success": False,
            "error": str(e)
        }
```

#API Function 4 - Log Email
This simulates a logging API

```python
def log_email(status, message):

    try:
        with open("email_log.txt", "a") as file:
            file.write(f"{status}: {message}\n")

        return {
            "success": True,
            "message": "Action logged successfully."
        }

    except Exception as e:
        return {
            "success": False,
            "error": str(e)
        }
```



---




#Automated Workflow
The system automatically calls the API-like functions in sequence

```python
customer_message = input("Enter customer message: ")

# 1. Generate reply
reply_result = generate_reply(customer_message)

if not reply_result["success"]:
    print("Error:", reply_result["error"])

else:

    email = reply_result["email"]

    print("\n--- GENERATED EMAIL ---")
    print(email)

    # 2. Save email
    save_result = save_email(email)

    print("\nSave API:")
    print(save_result)

    if not save_result["success"]:
        log_email("ERROR", "Failed to save email.")

    else:

        # 3. Send email
        send_result = send_email(email)

        print("\nSend API:")
        print(send_result)

        # 4. Log result
        if send_result["success"]:
            log_result = log_email(
                "SUCCESS",
                "Email generated, saved, and sent."
            )
        else:
            log_result = log_email(
                "ERROR",
                "Email sending failed."
            )

        print("\nLog API:")
        print(log_result)
```
